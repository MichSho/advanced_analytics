```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 2000

pre_spend = np.random.normal(50, 10, n)
treatment = np.random.binomial(1, 0.5, n)

# baseline relationship
noise = np.random.normal(0, 8, n)

# treatment effect = +5
exp_spend = 0.7 * pre_spend + 5 * treatment + noise

df = pd.DataFrame({
    "treatment": treatment,
    "pre_spend": pre_spend,
    "exp_spend": exp_spend
})
```

```python
cov_xy = np.cov(df["pre_spend"], df["exp_spend"])[0,1]
var_x = np.var(df["pre_spend"])

theta = cov_xy / var_x
print(theta)
```

```python
x_mean = df["pre_spend"].mean()

df["exp_spend_cuped"] = (
    df["exp_spend"] 
    - theta * (df["pre_spend"] - x_mean)
)
```

```python
print("Original variance:", df["exp_spend"].var())
print("CUPED variance:", df["exp_spend_cuped"].var())
```

```python
control = df[df.treatment == 0]["exp_spend_cuped"]
treat = df[df.treatment == 1]["exp_spend_cuped"]

effect = treat.mean() - control.mean()

print("Estimated treatment effect:", effect)
```
